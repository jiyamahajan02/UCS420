```python

r = 1024170305
L = [int(x) * 10 for x in str(r)]
print("1. L =",L)

L.append(80)
print("2. After append:",L)

L.insert(2, 65)
print("3. After insert:", L)

L.remove(70)
print("4. After remove:", L)

L.pop(1)
print("4. After pop:", L)

L.sort()
print("5. Ascending:", L)

L.sort(reverse=True)
print("5. Descending:",L)

print("6. First 3:",L[:3])
print("6. Last 3:", L[-3:])

a=sum(L) /len(L)
N=[x for x in L if x > a]

print("7. Average:", a)
print("Elements greater than average:", N)
```

```python
scores = tuple(L[:8])

h = max(scores)
hi = scores.index(h)
l = min(scores)
lc = scores.count(l)

print("Highest:", h, "Index:", hi)
print("Lowest:", l, "Count:", lc)

r=list(scores[::-1])
print("Reversed list:", r)

x=int(input("Enter a score: "))

if x in scores:
    print("First occurrence index:", scores.index(x))
else:
    print("Score not present")

f, s, *rem = scores
print("First:", f)
print("Second:", s)
print("Remaining:", rem)
```

```python
import random

random.seed(1024170305)
L = [random.randint(100, 900) for _ in range(100)]
print("1. Random numbers:", L)

o=[x for x in L if x%2 != 0]
print("2. Odd count:", len(o))
print("Odd numbers:", o)

e=[x for x in L if x%2 ==0]
print("3. Even count:", len(e))
print("Even numbers:", e)

def prime(x):
    if x < 2:
        return False
    for i in range(2, int(x ** 0.5)+1):
        if x % i == 0:
            return False
    return True

p = [x for x in L if prime(x)]
print("4. Prime count:", len(p))
print("Prime numbers:", p)

m = max(L, key=L.count)
print("5. Most frequent:", m)
print("Frequency:", L.count(m))
```

```python
d= [1,0,2,4,1,7,0,3]

A= {x*7 for x in d}
B={x *9 for x in d}

print("A:", A)
print("B:", B)

u=A.union(B)
print("Union:", u)

i=A.intersection(B)
print("Intersection:", i)

ab=A.difference(B)
ba=B.difference(A)

print("A - B:", ab)
print("B - A:", ba)

s=A.symmetric_difference(B)
print("Symmetric difference:", s)

print("A subset of B:", A.issubset(B))
print("B superset of A:", B.issuperset(A))

x=int(input("Enter X: "))
A.discard(x)

print("A after discard:", A)
```

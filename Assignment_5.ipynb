## Q2


```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("Reversed:", arr[::-1])

def most_frequent_and_indices(array):
    values, counts = np.unique(array, return_counts=True)
    maximum = counts.max()
    frequent = values[counts == maximum]
    return [(value, np.flatnonzero(array == value)) for value in frequent]

x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1,1,1,2,3,4,2,4,3,3])
print("x:", most_frequent_and_indices(x))
print("y:", most_frequent_and_indices(y))
```

## Q4


```python
jiya_mahajan = np.linspace(10, 100, 25)
print(jiya_mahajan)
print("Dimensions:", jiya_mahajan.ndim)
print("Shape:", jiya_mahajan.shape)
print("Elements:", jiya_mahajan.size)
print("Data type:", jiya_mahajan.dtype)
print("Bytes:", jiya_mahajan.nbytes)
transposed_by_reshape = jiya_mahajan.reshape(1, -1)
print("Row form using reshape:\n", transposed_by_reshape)
print("Column form using reshape:\n", jiya_mahajan.reshape(-1, 1))
print("T attribute on 1-D array:", jiya_mahajan.T.shape)
print("T attribute after making a 2-D row:\n", transposed_by_reshape.T)
```

## Q1


```python
import numpy as np
temperature = np.array([25,28,31,35,38,27,33,40])
corrected = temperature + 2
fahrenheit = corrected * 9 / 5 + 32
above_32 = corrected[corrected > 32]
print("Corrected Celsius:", corrected)
print("Fahrenheit:", fahrenheit)
print("Corrected readings above 32°C:", above_32)
print("Count:", above_32.size)
```

Vectorization applies compiled array operations to all values at once, avoiding Python-level loop overhead. Boolean indexing expresses a condition as an array mask and selects matching readings directly.

## Q2


```python
steps = np.array([[5000,6200,7100],[8000,7500,9000],[4500,5100,4800],[9000,8500,9500]])
print("Total:", steps.sum())
print("Mean:", steps.mean())
print("Maximum:", steps.max())
print("Minimum:", steps.min())
print("Totals per day:", steps.sum(axis=0))
print("Totals per user:", steps.sum(axis=1))
max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum at (user index, day index):", max_position)
```

## Q4


```python
X = np.array([[6,70,3],[5,50,6],[8,80,2],[4,30,8]], dtype=float)
y = np.array([40,65,30,85], dtype=float)
print("Shape:", X.shape, "dimensions:", X.ndim)
print("Transpose:\n", X.T)
xtx = X.T @ X
print("X.T @ X:\n", xtx)
print("Inverse:\n", np.linalg.inv(xtx))
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("Coefficients (sleep, activity, stress):", beta)
new_user = np.array([5,40,7])
print("Predicted score:", new_user @ beta)
```

The coefficients give the fitted change in predicted assistance score per one-unit change in sleep hours, activity level, and stress level, holding the other features fixed. The model has no intercept because the assignment’s equation does not include one.
